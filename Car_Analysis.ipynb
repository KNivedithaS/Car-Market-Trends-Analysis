```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("car_Dekho_DA.csv")
data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
print(data.Car_Name.value_counts())
```

```python
plt.figure(figsize=(12,20))
sns.countplot(y=data.Car_Name)
plt.show()
```

```python
print(data.Fuel_Type.value_counts())
```

```python
data.Fuel_Type.value_counts().plot.pie(
    autopct="%1.1f%%",
    figsize=(6,6)
)
plt.title("Fuel Type Distribution")
plt.ylabel("")
plt.show()
```

```python
data.Fuel_Type.value_counts().plot.bar()

plt.title("Vehicles by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Vehicles")
plt.show()
```

```python
top10 = data.Car_Name.value_counts().head(10)
top10.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Most Sold Vehicles")
plt.xlabel("Vehicle")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.show()
```

```python
data["Age"] = data.Year.max() - data.Year
plt.figure(figsize=(8,5))
sns.scatterplot(
    x=data.Age,
    y=data.Selling_Price
)
plt.title("Vehicle Age vs Selling Price")
plt.xlabel("Vehicle Age")
plt.ylabel("Selling Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x=data.Kms_Driven,
    y=data.Selling_Price
)
plt.title("Kilometers Driven vs Selling Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price")
plt.show()
```
